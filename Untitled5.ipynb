```python
import numpy as np
data = np.random.exponential(1, size=20)
data
```

```python
data.mean()
```

```python
data.var()
```

```python
import pandas as pd
import altair as alt

# 1. Crear un DataFrame con los datos observados
df_datos = pd.DataFrame({'valores': data})

# 2. Generar puntos para la curva de la distribución exponencial teórica
# Como usamos un parámetro de escala beta=1 (parámetro de tasa lambda=1/beta=1),
# la función de densidad de probabilidad (PDF) es f(x) = e^(-x) para x >= 0.
x_teorico = np.linspace(0, data.max() + 0.5, 200)
y_teorico = np.exp(-x_teorico)
df_teorico = pd.DataFrame({'x': x_teorico, 'y': y_teorico})

# 3. Crear el histograma de los datos empíricos (normalizado para mostrar densidad)
histograma = alt.Chart(df_datos).mark_bar(opacity=0.6, color='skyblue').encode(
    alt.X('valores:Q', bin=alt.Bin(maxbins=10), title='Valores'),
    alt.Y('density:Q', title='Densidad')
)

# 4. Crear la línea que representa la distribución teórica
linea_teorica = alt.Chart(df_teorico).mark_line(color='red', strokeWidth=2).encode(
    x='x:Q',
    y='y:Q'
)

# 5. Combinar ambos gráficos
grafico_final = (histograma + linea_teorica).properties(
    title='Datos Generados vs. Distribución Exponencial Teórica (Beta=1)',
    width=500,
    height=300
)

grafico_final
```

```python
import scipy.stats as stats
import numpy as np

# Parámetros basados en tu variable real 'data'
n = len(data)
alpha = 0.05
grados_libertad = n - 1

# Varianza muestral (corregido de '52' a 's2')
s2 = np.var(data, ddof=1)

# Valores críticos de chi-cuadrado
chi_inferior = stats.chi2.ppf(alpha / 2, grados_libertad)
chi_superior = stats.chi2.ppf(1 - alpha / 2, grados_libertad)

# Intervalo de confianza para la varianza poblacional
limite_inferior = (grados_libertad * s2) / chi_superior
limite_superior = (grados_libertad * s2) / chi_inferior

# Resultados corregidos con llaves para f-strings
print(f"Tamaño de la muestra: {n}")
print(f"Varianza muestral (s2): {s2:.4f}")
print(f"Valor crítico Chi-cuadrado inferior: {chi_inferior:.4f}")
print(f"Valor crítico Chi-cuadrado superior: {chi_superior:.4f}")
print(f"Intervalo de confianza del 95% para la varianza:")
print(f"[{limite_inferior:.4f}, {limite_superior:.4f}]")
```
