```python
## Simulación del TCL (DeMoivre-Laplace).
* Tomar muestras cierto tamaño.
* Media de las muestras
*. Histograma (representar a la normalidad)



```

```python
import numpy as npr
npr.random.seed(1)
muestras_binomial = npr.random.binomial(3, 0.6, 1000)
muestras_binomial
```

```python
import pandas as pd
import numpy.random as npr

df = pd.DataFrame()

for i in range(1, 101):
    muestras_binomiales = npr.binomial(1000, 0.6, 50)
    col = f"muestras_{i}"
    df[col] = muestras_binomiales

df_media = pd.DataFrame(df.mean(), columns= ["muestra de las medias"])
df_media
```

```python
import seaborn as sns
sns.distplot(df_media)
```

```python
import seaborn as sns
df = pd.DataFrame()

for i in range(1, 101):
    muestras_exponenciales = npr.exponential(1000,50)
    col = f"muestras_{i}"
    df[col] = muestras_exponenciales

df_media = pd.DataFrame(df.mean(), columns= ["muestra de las medias"])

sns.distplot(df_media)
```
